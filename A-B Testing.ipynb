```python
import numpy as np

# 2. Generating sample data for two versions

# Data for version A
version_A = np.random.normal(loc=10, scale=2, size=1000)  

# Data for version B
version_B = np.random.normal(loc=12, scale=2, size=1000)
```

```python
from scipy.stats import ttest_ind

# 3. Performing the t-test to compare the means
t, p = ttest_ind(version_A, version_B)  # t: test statistic, p: p-value

# 4. Printing the results
print(f"t = {t:.3f}, p = {p:.3f}")
```

```python
import matplotlib.pyplot as plt

# 5. Plotting histograms and means for each version

# Histogram for version A
plt.hist(version_A, alpha=0.5, label='Version A')

# Histogram for version B  
plt.hist(version_B, alpha=0.5, label='Version B')  

# Mean line for version A
plt.axvline(np.mean(version_A), color='r', linestyle='dashed', linewidth=1, label='Mean A')  

# Mean line for version B
plt.axvline(np.mean(version_B), color='b', linestyle='dashed', linewidth=1, label='Mean B')  

# Legend in the upper right corner
plt.legend(loc='upper right')  

# X-axis label
plt.xlabel('Values') 

# Y-axis label 
plt.ylabel('Frequency')  

# Plot title
plt.title('A/B Test Results')  

# Displaying the plot
plt.show() 
```
