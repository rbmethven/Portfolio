Suggested metrics to get started:

Revenue Generated (RG)
Opportunity Conversion Rate (OCR)
Average Deal Size (ADS)
Customer Retention Rate (CRR)
Number of New Customers Acquired (NCA)
The equation for sWRC+ could be:

sWRC+ = (RG * 0.4) + (OCR * 20) + (ADS * 0.00001) + (CRR * 30) + (NCA * 0.5)

This equation weights each metric based on its perceived importance. The weights can be adjusted based on specific business priorities.

Here's a Python implementation that calculates this sWRC+ for a group of sales people:

## Linear

```python
import pandas as pd
import numpy as np

# Sample data
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Revenue_Generated': [1000000, 800000, 1200000, 950000, 1100000],
    'Opportunity_Conversion_Rate': [0.3, 0.25, 0.35, 0.28, 0.32],
    'Average_Deal_Size': [50000, 40000, 60000, 47500, 55000],
    'Customer_Retention_Rate': [0.9, 0.85, 0.92, 0.88, 0.91],
    'New_Customers_Acquired': [20, 15, 25, 18, 22]
})

# Normalize the data
for col in ['Revenue_Generated', 'Opportunity_Conversion_Rate', 'Average_Deal_Size', 'Customer_Retention_Rate', 'New_Customers_Acquired']:
    data[col] = (data[col] - data[col].min()) / (data[col].max() - data[col].min())

# Calculate sWRC+
data['sWRC+'] = (
    (data['Revenue_Generated'] * 0.4) +
    (data['Opportunity_Conversion_Rate'] * 20) +
    (data['Average_Deal_Size'] * 0.00001) +
    (data['Customer_Retention_Rate'] * 30) +
    (data['New_Customers_Acquired'] * 0.5)
)

# Calculate league average sWRC+
league_avg_swrc = data['sWRC+'].mean()

# Normalize sWRC+ around league average (100 is average)
data['sWRC+'] = (data['sWRC+'] / league_avg_swrc) * 100

# Sort by sWRC+ in descending order
result_sorted = data.sort_values('sWRC+', ascending=False).reset_index(drop=True)

print(result_sorted)
```

## With Functions

```python
import pandas as pd
import numpy as np

def normalize_data(data):
    """
    Normalize the input data to a 0-1 scale.
    
    :param data: DataFrame containing performance metrics
    :return: DataFrame with normalized metrics
    """
    for col in ['Revenue_Generated', 'Opportunity_Conversion_Rate', 'Average_Deal_Size', 'Customer_Retention_Rate', 'New_Customers_Acquired']:
        data[col] = (data[col] - data[col].min()) / (data[col].max() - data[col].min())
    return data

def calculate_swrc_plus(data):
    """
    Calculate sWRC+ for each sales person based on their normalized performance metrics.
    
    :param data: DataFrame containing normalized performance metrics
    :return: DataFrame with original data and calculated sWRC+
    """
    # Calculate sWRC+
    data['sWRC+'] = (
        (data['Revenue_Generated'] * 0.4) +
        (data['Opportunity_Conversion_Rate'] * 20) +
        (data['Average_Deal_Size'] * 0.00001) +
        (data['Customer_Retention_Rate'] * 30) +
        (data['New_Customers_Acquired'] * 0.5)
    )
    
    # Calculate league average sWRC+
    league_avg_swrc = data['sWRC+'].mean()
    
    # Normalize sWRC+ around league average (100 is average)
    data['sWRC+'] = (data['sWRC+'] / league_avg_swrc) * 100
    
    return data

# Sample data
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Revenue_Generated': [1000000, 800000, 1200000, 950000, 1100000],
    'Opportunity_Conversion_Rate': [0.3, 0.25, 0.35, 0.28, 0.32],
    'Average_Deal_Size': [50000, 40000, 60000, 47500, 55000],
    'Customer_Retention_Rate': [0.9, 0.85, 0.92, 0.88, 0.91],
    'New_Customers_Acquired': [20, 15, 25, 18, 22]
})

# Normalize the data
data = normalize_data(data)

# Calculate sWRC+
result = calculate_swrc_plus(data)

# Sort by sWRC+ in descending order
result_sorted = result.sort_values('sWRC+', ascending=False).reset_index(drop=True)

print(result_sorted)
```

## With Industry and geo adjustments

```python
import pandas as pd
import numpy as np

# Sample data
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Revenue_Generated': [1000000, 800000, 1200000, 950000, 1100000],
    'Opportunity_Conversion_Rate': [0.3, 0.25, 0.35, 0.28, 0.32],
    'Average_Deal_Size': [50000, 40000, 60000, 47500, 55000],
    'Customer_Retention_Rate': [0.9, 0.85, 0.92, 0.88, 0.91],
    'New_Customers_Acquired': [20, 15, 25, 18, 22],
    'Industry': ['Financial', 'Retail', 'Healthcare', 'Technology', 'Manufacturing'],
    'Region': ['EMEA', 'NA', 'APAC', 'NA', 'EMEA']
})

# Industry adjustment factors (based on historical difficulty of selling in each industry)
industry_factors = {
    'Financial': 1.2,      # Financial services are harder to sell to (regulated)
    'Retail': 0.9,        # Retail might be easier to sell to
    'Healthcare': 1.3,     # Healthcare is highly regulated and complex
    'Technology': 0.8,     # Tech companies might be more receptive to AWS
    'Manufacturing': 1.1   # Manufacturing has moderate complexity
}

# Regional adjustment factors (based on market maturity and competition)
region_factors = {
    'NA': 0.9,    # North America is a mature market
    'EMEA': 1.1,  # Europe/Middle East/Africa has more complex requirements
    'APAC': 1.2   # Asia Pacific might be a developing market with unique challenges
}

# Add adjustment columns
data['Industry_Factor'] = data['Industry'].map(industry_factors)
data['Region_Factor'] = data['Region'].map(region_factors)

# Normalize the core metrics
metrics_to_normalize = ['Revenue_Generated', 'Opportunity_Conversion_Rate', 
                       'Average_Deal_Size', 'Customer_Retention_Rate', 
                       'New_Customers_Acquired']

for col in metrics_to_normalize:
    data[col] = (data[col] - data[col].min()) / (data[col].max() - data[col].min())

# Calculate base sWRC+
data['Base_sWRC+'] = (
    (data['Revenue_Generated'] * 0.4) +
    (data['Opportunity_Conversion_Rate'] * 0.2) +
    (data['Average_Deal_Size'] * 0.15) +
    (data['Customer_Retention_Rate'] * 0.15) +
    (data['New_Customers_Acquired'] * 0.1)
)

# Apply adjustments to get adjusted sWRC+
data['Adjusted_sWRC+'] = data['Base_sWRC+'] * data['Industry_Factor'] * data['Region_Factor']

# Normalize around 100 as average
league_avg_swrc = data['Adjusted_sWRC+'].mean()
data['Final_sWRC+'] = (data['Adjusted_sWRC+'] / league_avg_swrc) * 100

# Sort by final sWRC+ in descending order
result_sorted = data.sort_values('Final_sWRC+', ascending=False).reset_index(drop=True)

# Select columns to display
columns_to_display = ['Name', 'Industry', 'Region', 'Base_sWRC+', 
                     'Industry_Factor', 'Region_Factor', 'Final_sWRC+']

print(result_sorted[columns_to_display].round(2))

# Calculate and print average sWRC+ by industry and region
print("\nAverage sWRC+ by Industry:")
print(data.groupby('Industry')['Final_sWRC+'].mean().round(2))

print("\nAverage sWRC+ by Region:")
print(data.groupby('Region')['Final_sWRC+'].mean().round(2))
```
